# TP1: Linear Regression (Simple & Multiple)
**University Assignment**: Linear Regression on Diabetes and Iris Datasets.
Colab-ready notebook using scikit-learn, NumPy, Pandas, and Matplotlib.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_diabetes, load_iris
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
print('Libraries successfully imported!')
```

## Part A: Diabetes Dataset
- Simple Linear Regression (SLR): BMI feature (index 2) -> target
- Multiple Linear Regression (MLR): all 10 features -> target
- train_test_split(test_size=0.2, random_state=42)

```python
# Load Diabetes
diabetes = load_diabetes()
X_d = diabetes.data
y_d = diabetes.target

X_tr_d, X_te_d, y_tr_d, y_te_d = train_test_split(
    X_d, y_d, test_size=0.2, random_state=42
)

# 1. Simple Linear Regression (BMI = col index 2)
slr_d = LinearRegression()
slr_d.fit(X_tr_d[:, [2]], y_tr_d)
y_pred_slr_d = slr_d.predict(X_te_d[:, [2]])

mae_slr_d = mean_absolute_error(y_te_d, y_pred_slr_d)
mse_slr_d = mean_squared_error(y_te_d, y_pred_slr_d)
rmse_slr_d = np.sqrt(mse_slr_d)
r2_slr_d = r2_score(y_te_d, y_pred_slr_d)

# 2. Multiple Linear Regression (All 10 features)
mlr_d = LinearRegression()
mlr_d.fit(X_tr_d, y_tr_d)
y_pred_mlr_d = mlr_d.predict(X_te_d)

mae_mlr_d = mean_absolute_error(y_te_d, y_pred_mlr_d)
mse_mlr_d = mean_squared_error(y_te_d, y_pred_mlr_d)
rmse_mlr_d = np.sqrt(mse_mlr_d)
r2_mlr_d = r2_score(y_te_d, y_pred_mlr_d)

print(f'Diabetes SLR (BMI): MAE={mae_slr_d:.2f}, MSE={mse_slr_d:.1f}, RMSE={rmse_slr_d:.2f}, R2={r2_slr_d:.4f}')
print(f'Diabetes MLR:      MAE={mae_mlr_d:.2f}, MSE={mse_mlr_d:.1f}, RMSE={rmse_mlr_d:.2f}, R2={r2_mlr_d:.4f}')
```

## Theory Questions & Answers
1. **What is the difference between SLR and MLR?**
   - SLR models $y = ax + b$ with one input variable, while MLR models $y = \sum w_i x_i + b$ with multiple inputs, capturing multivariate interactions.
2. **Role of slope and intercept?**
   - Slope represents the rate of change of $y$ per unit increase of $x$. Intercept is the expected value of $y$ when all $x=0$.
3. **Perceptron vs Linear Regression?**
   - Linear Regression predicts continuous numerical targets via OLS (MSE minimization). Perceptron performs discrete classification with an activation threshold and error-based updates.
